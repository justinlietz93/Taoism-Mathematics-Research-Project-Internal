```python
import sympy as sp
import matplotlib.pyplot as plt
p11,p12,p21,p22,sigma=sp.symbols('p11 p12 p21 p22 sigma')
P=sp.Matrix([[p11,p12],[p21,p22]])
E=sp.Matrix([[p11,p12,0],[p21,p22,0],[0,0,sigma]])
ok=(E[:2,:2]==P and E[:2,2]==sp.zeros(2,1) and E[2,:2]==sp.zeros(1,2))
print(('CONDITIONAL CHECK' if ok else 'FAIL')+': two-sided P-orthogonal L birth gives block diagonal extension')
print('hypothesis=both mixed blocks are zero; extension=');sp.pprint(E)
fig,ax=plt.subplots();ax.imshow([[1,1,0],[1,1,0],[0,0,1]],vmin=0,vmax=1);ax.set(title='Conditional two-sided L block support');plt.show()
```

```python
import sympy as sp
import matplotlib.pyplot as plt
u,v=sp.symbols('u v',positive=True)
ratio=sp.simplify((1/(v*(u+v)))/(1/(u*v)))
ok=sp.simplify(ratio-u/(u+v))==0
print(('PASS' if ok else 'FAIL')+': local B descendant a -> a*u/(u+v)')
print('exact ratio=',ratio)
vals=[float(sp.Rational(n,2*n+1)) for n in range(1,8)]
fig,ax=plt.subplots();ax.plot(range(1,8),vals,marker='o');ax.set(title='Local B scale factor samples',xlabel='u, with v=u+1',ylabel='u/(u+v)');plt.show()
```

```python
from fractions import Fraction
import matplotlib.pyplot as plt
word='BQQBBBQBQBBQBBL';u=v=1;q=0;values=[complex(1,0)]
for op in word:
    if op=='B':u,v=v,u+v
    elif op=='Q':q=(q+1)%4
    values.append((1j**q)/(u*v))
endpoint=values[-2]
ok=((u,v)==(55,89) and q==1 and abs(endpoint-1j/4895)<1e-18)
print(('PASS' if ok else 'FAIL')+': recursive first-domain interpreter reaches i/4895 before L')
print('pair=',(u,v),'qmod=',q,'endpoint=',endpoint)
fig,ax=plt.subplots();ax.plot(range(len(values)),[z.real for z in values],label='Re');ax.plot(range(len(values)),[z.imag for z in values],label='Im');ax.legend();ax.set(title='First-domain local active-axis trace',xlabel='prefix index');plt.show()
```

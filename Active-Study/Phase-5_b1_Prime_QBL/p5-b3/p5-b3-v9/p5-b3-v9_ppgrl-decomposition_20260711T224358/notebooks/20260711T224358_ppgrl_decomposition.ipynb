```python
import sympy as sp
import matplotlib.pyplot as plt
from fractions import Fraction

word='BQQBBBQBQBBQBBL'
u,v=1,1; q=0; trace=[]
for op in word:
    if op=='B': u,v=v,u+v
    elif op=='Q': q+=1
    phase=[(1,0),(0,1),(-1,0),(0,-1)][q%4]
    trace.append((op,u,v,q,phase))
assert (u,v,q)==(55,89,5)
print('PASS: exact Domain-0 custody prefix')
print('endpoint pair =', (u,v), 'Q count =', q)
fig,ax=plt.subplots(); ax.plot(range(len(trace)),[a*b for _,a,b,_,_ in trace],marker='o'); ax.set_yscale('log'); ax.set_title('Domain-0 pair product'); plt.show()
```

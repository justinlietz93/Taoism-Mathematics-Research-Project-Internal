```python

import math, sympy as sp, mpmath as mp, numpy as np, matplotlib.pyplot as plt
mp.mp.dps=100
phi=(1+mp.sqrt(5))/2
lam=mp.log(2)/mp.log(phi)
beta=mp.log(5)/(2*mp.log(phi))-mp.mpf(3)/2
gamma=6*lam-beta
def threshold(A):
    y=6*lam*(2**(A+1)-1)+beta
    return int(mp.ceil(y))
def fib(n):
    if n==0:return 0
    a,b=0,1
    for _ in range(n):a,b=b,a+b
    return a
def finish(title, text):
    fig,ax=plt.subplots(figsize=(7,3)); ax.axis('off'); ax.text(0.02,0.95,title,va='top',fontsize=12); ax.text(0.02,0.78,text,va='top',family='monospace',fontsize=9); plt.show()

A=sp.symbols('A',integer=True,nonnegative=True)
J=6*(2**(A+1)-1)
checks=[sp.simplify(6*(2**(A+2)-1)-(2*J+6)),sp.simplify(12*(2**(A+1)-1)-2*J)]
ok=checks==[0,0]
print(('PASS' if ok else 'FAIL')+': exact boundary and capacity identities')
print('J_(A+1)=2J_A+6; m_A=2J_A')
finish('Claim 1: custody index identities',str(checks))
```

```python

lam_s,beta_s,j_s,b_s,c_s=sp.symbols('lambda beta j b c',real=True)
gamma_s=6*lam_s-beta_s
res=sp.expand(lam_s*(2*j_s+6)+beta_s-(2*b_s+c_s)-(2*(lam_s*j_s+beta_s-b_s)+gamma_s-c_s))
print(('PASS' if res==0 else 'FAIL')+': exact semiconjugacy algebra')
print('symbolic residual =',res)
finish('Claim 2: factor algebra',f'residual = {res}')
```

```python

rows=[]
for A0 in range(9):
    t=threshold(A0); rows.append((A0,t,fib(t+1),fib(t+2),6*(2**(A0+1)-1)))
ok=all(u==fib(t+1) and v==fib(t+2) for A0,t,u,v,jj in rows)
print(('PASS' if ok else 'FAIL')+': finite exact boundary regression A=0..8')
print(rows[:4])
fig,ax=plt.subplots(figsize=(7,3)); ax.plot([r[0] for r in rows],[r[1] for r in rows],marker='o'); ax.set_xlabel('A'); ax.set_ylabel('T_A'); ax.set_title('Finite regression only: threshold counts'); plt.show()
```

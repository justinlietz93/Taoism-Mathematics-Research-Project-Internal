# QBL Global Exact-Threshold Bridge

No file I/O. Each code cell states one claim, prints PASS/FAIL, and emits one figure.

```python
import math
import hashlib
import sympy as sp
import mpmath as mp
import matplotlib.pyplot as plt
mp.mp.dps = 100
phi = (mp.mpf(1)+mp.sqrt(5))/2
lnphi = mp.log(phi)
ln2 = mp.log(2)
ln5 = mp.log(5)

# Claim 1: affine constants and recurrence
alpha = 6*ln2/lnphi
gamma = alpha + mp.mpf('1.5') - ln5/(2*lnphi)
def mA(A): return 12*(2**(A+1)-1)
def yA(A): return (mA(A)*ln2+ln5)/(2*lnphi)-mp.mpf('1.5')
errs=[abs(yA(A)-(2*yA(A-1)+gamma)) for A in range(1,20)]
ok=max(errs)<mp.mpf('1e-90')
print(('PASS' if ok else 'FAIL'), 'alpha=', mp.nstr(alpha,30), 'gamma=', mp.nstr(gamma,30), 'max residual=', mp.nstr(max(errs),5))
fig,ax=plt.subplots(); ax.plot(range(1,20), [float(e) for e in errs], marker='o'); ax.set_yscale('log'); ax.set_title('Affine recurrence residual'); ax.set_xlabel('A'); ax.set_ylabel('absolute residual'); plt.show()
```

```python
import math
import hashlib
import sympy as sp
import mpmath as mp
import matplotlib.pyplot as plt
mp.mp.dps = 100
phi = (mp.mpf(1)+mp.sqrt(5))/2
lnphi = mp.log(phi)
ln2 = mp.log(2)
ln5 = mp.log(5)

# Claim 4: equality obstruction begins and ends at products 1 and 2
def fib_pair(n):
    if n==0: return (0,1)
    a,b=fib_pair(n//2); c=a*(2*b-a); d=a*a+b*b
    return (d,c+d) if n%2 else (c,d)
p=[]
for n0 in range(30):
    a,b=fib_pair(n0+1); P=a*b
    ispow=P>0 and (P & (P-1))==0
    if ispow: p.append((n0,P))
ok=(p==[(0,1),(1,2)])
print(('PASS' if ok else 'FAIL'), 'power-of-two consecutive products=', p, 'all QBL X_A >= 4096')
fig,ax=plt.subplots(); ax.scatter([x for x,_ in p],[y for _,y in p]); ax.set_yscale('log'); ax.set_title('Power-of-two consecutive Fibonacci products'); ax.set_xlabel('n'); ax.set_ylabel('P_n'); plt.show()
```

# Phase 5 v7w: Parity Seating / Post-L Mod-12 Reverification

No file IO. Each code cell computes a claim directly.

```python
from math import gcd
import matplotlib.pyplot as plt

def chi12(n):
    r = n % 12
    return 1 if r in (1,11) else (-1 if r in (5,7) else 0)
def seat6(n): return n % 6
def parity_latch(n): return (n % 12)//6
def seat12(n): return seat6(n)+6*parity_latch(n)
rows = list(range(12))
vals = [seat12(r) for r in rows]
passed = all(seat12(r)==r for r in rows)
plt.figure()
plt.bar(rows, vals)
plt.title('Claim 1: post-L latch reconstructs mod-12 seating')
plt.xlabel('residue mod 12')
plt.ylabel('post-L seat')
plt.show()
print('PASS' if passed else 'FAIL', {'checks': len(rows), 'failures': [r for r in rows if seat12(r)!=r]})
```

```python
pairs = [(1,7),(5,11)]
passed = all(seat6(a)==seat6(b) and chi12(a)!=chi12(b) for a,b in pairs)
plt.figure()
plt.bar([f'{a},{b}' for a,b in pairs], [chi12(a)-chi12(b) for a,b in pairs])
plt.title('Claim 2: mod-6 collision is a chi12 obstruction')
plt.xlabel('pre-L collision pair')
plt.ylabel('chi12(a)-chi12(b)')
plt.show()
print('PASS' if passed else 'FAIL', {'pairs': pairs, 'pre_L_seats': [(seat6(a),seat6(b)) for a,b in pairs]})
```

```python
support_terms = [n for n in range(1,49) if gcd(n,6)==1]
passed = all(chi12(n)==chi12(seat12(n)) for n in support_terms)
plt.figure()
plt.plot(support_terms, [chi12(n) for n in support_terms], marker='o')
plt.title('Claim 3: chi12 recovered from post-L seat on support terms')
plt.xlabel('n')
plt.ylabel('chi12')
plt.show()
print('PASS' if passed else 'FAIL', {'support_terms': len(support_terms), 'mismatches': [n for n in support_terms if chi12(n)!=chi12(seat12(n))]})
```

```python
controls = {
    'drop_latch_rejected': seat6(1)==seat6(7) and chi12(1)!=chi12(7),
    'parity_without_base_rejected': parity_latch(1)==parity_latch(5) and chi12(1)!=chi12(5),
    'supportless_zero': all(chi12(r)==0 for r in [0,2,3,4,6,8,9,10]),
}
passed = all(controls.values())
plt.figure()
plt.bar(list(controls.keys()), [int(v) for v in controls.values()])
plt.title('Claim 4: negative controls')
plt.xticks(rotation=20, ha='right')
plt.ylim(0,1.2)
plt.show()
print('PASS' if passed else 'FAIL', controls)
```

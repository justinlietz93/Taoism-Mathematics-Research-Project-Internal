## SymPy check: gauge action preserves determinant unit status


```python
N = 8
S = sp.Matrix([[0,1],[1,0]])
P = sp.Matrix([[1,1],[0,1]])
S2 = (P.T*S*P) % N
detS = int(S.det()) % N
detS2 = int(S2.det()) % N
unit_before = sp.gcd(detS, N) == 1
unit_after = sp.gcd(detS2, N) == 1

plt.figure(figsize=(4,3))
plt.imshow([[int(x) for x in row] for row in S2.tolist()])
plt.title('Transformed form entries mod 8')
plt.xticks([0,1]); plt.yticks([0,1])
plt.tight_layout()
plt.show()

print('PASS' if unit_before and unit_after else 'FAIL', {'detS_mod_N': detS, 'detS2_mod_N': detS2, 'unit_before': unit_before, 'unit_after': unit_after, 'S2': S2.tolist()})
```

## Negative control: degenerate radical is rejected


```python
N = 6
D = sp.Matrix([[1,0],[0,0]])
radical = []
for x in range(N):
    for y in range(N):
        v = sp.Matrix([x,y])
        if all(int((v.T*D*sp.Matrix([z,w]))[0]) % N == 0 for z,w in [(1,0),(0,1)]):
            radical.append((x,y))

plt.figure(figsize=(4,3))
plt.scatter([p[0] for p in radical], [p[1] for p in radical])
plt.title('Degenerate radical points mod 6')
plt.xlabel('x'); plt.ylabel('y')
plt.tight_layout()
plt.show()

print('PASS' if len(radical) > 1 else 'FAIL', {'radical_size': len(radical), 'radical': radical[:10]})
```

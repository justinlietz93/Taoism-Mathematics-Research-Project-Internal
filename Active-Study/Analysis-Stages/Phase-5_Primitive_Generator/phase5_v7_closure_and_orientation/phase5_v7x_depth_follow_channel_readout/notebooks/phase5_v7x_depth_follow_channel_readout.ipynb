# Phase 5 v7x: Depth 3-6 Follow Channel-Field Readout

```
import math
import matplotlib.pyplot as plt

def chi12(n):
    r=n%12
    return 1 if r in (1,11) else -1 if r in (5,7) else 0
def support(n): return math.gcd(n,6)==1
def post_l_seat(n): return n%6 + 6*((n%12)//6)
def lap_sign(lap,n): return chi12(n) if lap%2==1 else -chi12(n)
def terms(depth): return [n for n in range(1,2**(depth+2)) if support(n)]

counts=[len(terms(d)) for d in range(3,7)]
passed=all(counts[i]<counts[i+1] for i in range(len(counts)-1))
plt.figure()
plt.plot([3,4,5,6], counts, marker='o')
plt.title('Support terms by depth')
plt.xlabel('depth')
plt.ylabel('support term count')
plt.show()
print('CLAIM depth expansion emits increasing support apertures')
print('counts=', counts)
print('PASS' if passed else 'FAIL')
```

```
depths=[3,4,5,6]
fail=[]
for d in depths:
    for n in terms(d):
        if lap_sign(2,n) != -lap_sign(1,n):
            fail.append((d,n))
plt.figure()
for d in depths:
    xs=terms(d)[:32]
    ys=[lap_sign(1,n)+lap_sign(2,n) for n in xs]
    plt.plot(xs, ys, marker='o', label=f'depth {d}')
plt.title('Lap antisymmetry residual: lap1 + lap2')
plt.xlabel('n')
plt.ylabel('residual')
plt.legend()
plt.show()
print('CLAIM lap-2 equals negative lap-1')
print('failures=', fail[:10], 'count=', len(fail))
print('PASS' if not fail else 'FAIL')
```

```
fail=[]
for d in [3,4,5,6]:
    for n in terms(d):
        if lap_sign(1,n) != chi12(n) or lap_sign(2,n) != -chi12(n):
            fail.append((d,n))
plt.figure()
xs=terms(4)[:40]
plt.plot(xs, [chi12(n) for n in xs], marker='o', label='chi12')
plt.plot(xs, [lap_sign(2,n) for n in xs], marker='x', label='lap2')
plt.title('Shadow sign channel and flipped lap-2 orientation')
plt.xlabel('n')
plt.ylabel('sign')
plt.legend()
plt.show()
print('CLAIM Follow channel matches external chi12 without scalar cargo')
print('failures=', fail[:10], 'count=', len(fail))
print('state_carries_scalar_residual=False')
print('PASS' if not fail else 'FAIL')
```

```
bad=[]
for d in [3,4,5]:
    if terms(d+1)[:len(terms(d))] != terms(d):
        bad.append(d)
plt.figure()
for d in [3,4,5,6]:
    xs=terms(d)[:24]
    plt.plot(range(len(xs)), [post_l_seat(n) for n in xs], marker='o', label=f'depth {d}')
plt.title('Post-L seat prefix stability')
plt.xlabel('term index')
plt.ylabel('post-L seat mod 12')
plt.legend()
plt.show()
print('CLAIM deeper apertures preserve prior readout prefix')
print('bad_depths=', bad)
print('PASS' if not bad else 'FAIL')
```

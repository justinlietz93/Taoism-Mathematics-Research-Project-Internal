```
import math, cmath
import matplotlib.pyplot as plt
N=12
chi=[0,1,0,0,0,-1,0,-1,0,0,0,1]
K=[[cmath.exp(-2j*math.pi*r*s/N)/math.sqrt(N) for r in range(N)] for s in range(N)]
F=[sum(K[s][r]*chi[r] for r in range(N)) for s in range(N)]
res=[abs(F[s]-chi[s]) for s in range(N)]
print('CLAIM: K_12 chi = chi')
print('max_residual =', max(res))
print('threshold =', 1e-12)
print('PASS' if max(res) <= 1e-12 else 'FAIL')
plt.figure()
plt.stem(range(N), res)
plt.title('Fourier fixed-vector residual by residue')
plt.xlabel('residue mod 12')
plt.ylabel('|K chi - chi|')
plt.show()
```

```
from fractions import Fraction
units=[1,5,7,11]
q={r: Fraction((r*r)%24,24) for r in units}
print('CLAIM: supported residues have locked fractional exponent 1/24')
print(q)
print('PASS' if all(v == Fraction(1,24) for v in q.values()) else 'FAIL')
plt.figure()
plt.bar([str(r) for r in units], [float(q[r]) for r in units])
plt.title('Fractional exponent channel on unit support')
plt.xlabel('residue mod 12')
plt.ylabel('q(r) mod 1')
plt.show()
```

```
def chi12(n):
    return {0:0,1:1,2:0,3:0,4:0,5:-1,6:0,7:-1,8:0,9:0,10:0,11:1}[n%12]
def support(n):
    return math.gcd(n,6)==1
terms=[]
n=1
while len(terms)<85:
    if support(n): terms.append(n)
    n += 1
mismatches=[n for n in terms if chi12(n)==0 or Fraction((n*n)%24,24) != Fraction(1,24)]
print('CLAIM: depth-6 aperture support terms match finite FQM support/sign/fractional channel')
print('checked_terms =', len(terms))
print('mismatch_count =', len(mismatches))
print('PASS' if not mismatches else 'FAIL')
plt.figure()
plt.plot(terms, [chi12(n) for n in terms], marker='o')
plt.title('Shadow sign channel over depth-6 support aperture')
plt.xlabel('n')
plt.ylabel('chi12(n)')
plt.show()
```

```
support_only=[0]*N
for r in [1,5,7,11]: support_only[r]=1
Fsup=[sum(K[s][r]*support_only[r] for r in range(N)) for s in range(N)]
res=max(abs(Fsup[s]-support_only[s]) for s in range(N))
print('NEGATIVE CONTROL: support-only vector is not the mock-theta FQM vector')
print('residual =', res)
print('threshold =', 1e-3)
print('PASS' if res > 1e-3 else 'FAIL')
plt.figure()
plt.stem(range(N), [abs(Fsup[s]-support_only[s]) for s in range(N)])
plt.title('Negative control residual: support-only vector')
plt.xlabel('residue mod 12')
plt.ylabel('residual')
plt.show()
```

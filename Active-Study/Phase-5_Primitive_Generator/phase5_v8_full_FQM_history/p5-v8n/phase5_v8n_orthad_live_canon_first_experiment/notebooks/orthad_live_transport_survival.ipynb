No file IO. Exact first-period data are constructed inline.

```python
import math
import matplotlib.pyplot as plt
from sympy import kronecker_symbol
ns = list(range(1, 13))
before = [int(kronecker_symbol(12, n)) for n in ns]
post_seats = [n % 12 for n in ns]
def seat_character(s):
    return 1 if s in (1, 11) else -1 if s in (5, 7) else 0
after = [seat_character(s) for s in post_seats]
plt.figure(figsize=(8, 3))
plt.plot(ns, before, marker="o", label="before")
plt.plot(ns, after, marker="x", label="after")
plt.xlabel("n")
plt.ylabel("character channel")
plt.title("Character channel across one cusp crossing")
plt.legend()
plt.show()
print("PASS" if before == after else "FAIL", {"survived": sum(a == b for a, b in zip(before, after)), "channels": 12})
```

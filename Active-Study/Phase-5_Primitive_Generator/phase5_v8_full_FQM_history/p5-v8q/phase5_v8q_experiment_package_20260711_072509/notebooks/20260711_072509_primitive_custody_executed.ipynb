# Canon First primitive custody attack

No experiment data files are read. Each claim is recomputed from the package source.

```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd() / 'src'))
import matplotlib
matplotlib.rcParams['svg.hashsalt'] = '20260711_072509'
import matplotlib.pyplot as plt
from sympy import Integer
from primitive_custody.application.engine import run_to_first_l_and_next_b
from primitive_custody.application.evidence import summarize
lifted, records = run_to_first_l_and_next_b()
result = summarize(records)
capacities = [Integer(r.capacity_before) for r in records]
plt.figure()
plt.plot(range(1, len(capacities)+1), capacities, marker='o')
plt.yscale('log', base=2)
plt.xlabel('primitive step')
plt.ylabel('capacity before step')
plt.title('Derived capacity sequence')
plt.show()
passed = result['crossing_word'] == 'BQQBBBQBQBBQBBL' and result['q_steps'] == 5
print('PASS' if passed else 'FAIL', {'word': result['crossing_word'], 'q_steps': result['q_steps'], 'rows': len(records)})
```

```python
products = [Integer(r.after['pair_product']) for r in records]
plt.figure()
plt.plot(range(1, len(products)+1), products, marker='o')
plt.yscale('log', base=2)
plt.xlabel('primitive step')
plt.ylabel('carried pair product')
plt.title('Pair carry through L')
plt.show()
passed = (result['floor_pair'] == [55,89] and result['post_l_pair'] == [55,89] and result['first_next_domain_pair'] == [89,144])
print('PASS' if passed else 'FAIL', {'floor_product': result['floor_product'], 'post_l_A': result['post_l_A'], 'post_l_k': result['post_l_k'], 'next_pair_product': 89*144})
```

# p5_v8v Pairing-First Orthad Realignment

No file I/O. One code cell per claim. Each claim emits one single-axes figure and prints the claim boundary.

```python
import matplotlib.pyplot as plt
A,u,v,ph,k,j,w=0,1,1,0,0,1,''
rows=[]
saw=False
for step in range(100):
    N=6*(2**A)
    cap=2 if j==1 else 4 if j==2 else 2**(2*j)
    nu,nv=v,u+v
    can_q=k<N-1
    can_b=(nu*nv<=cap) if can_q else (u*v<cap)
    p='B' if can_b else 'Q' if can_q else 'L'
    if p=='B': u,v=nu,nv
    elif p=='Q': ph+=1; k+=1; j+=1
    else: A+=1; k=0; j=1+6*((2**A)-1)
    w+=p; rows.append((p,w,A,u,v,ph,k,j))
    if p=='L': saw=True
    elif saw: break
word=rows[-2][1]
passed=word=='BQQBBBQBQBBQBBL'
print('PASS' if passed else 'FAIL')
print('exact values:', {'word':word,'rows':len(rows)})
print('claim boundary: primitive custody only')
fig,ax=plt.subplots()
ax.plot(range(1,len(word)+1), [1 if c=='B' else 2 if c=='Q' else 3 for c in word], marker='o')
ax.set_xlabel('primitive prefix length'); ax.set_ylabel('B=1 Q=2 L=3'); ax.set_title('Self-selected first crossing')
plt.show()
```

```python
lrow=rows[-2]
nextrow=rows[-1]
values={'before_L_pair':(55,89),'after_L':(lrow[2],lrow[3],lrow[4],lrow[5],lrow[6],lrow[7]),'after_next_B_pair':(nextrow[3],nextrow[4])}
passed=(lrow[2],lrow[3],lrow[4],lrow[5],lrow[6],lrow[7])==(1,55,89,5,0,7) and (nextrow[3],nextrow[4])==(89,144)
print('PASS' if passed else 'FAIL')
print('exact values:', values)
print('claim boundary: custody carry and first next-domain B')
fig,ax=plt.subplots()
ax.bar(['u before/after L','v before/after L','next u','next v'],[55,89,89,144])
ax.set_ylabel('exact integer'); ax.set_title('First-L carry and next refinement')
plt.show()
```

```python
phase=0; den=1; labels=[]
state_u,state_v=1,1
for p in 'BQQBBBQBQBBQBBL':
    if p=='B': state_u,state_v=state_v,state_u+state_v; den=state_u*state_v
    elif p=='Q': phase+=1
    labels.append((p,phase,den))
phase_label=('1','i','-1','-i')[phase%4]
shorthand=f'{phase_label}/{den}'
passed=shorthand=='i/4895'
print('PASS' if passed else 'FAIL')
print('exact values:', {'phase_quarters':phase,'denominator':den,'local_shorthand':shorthand})
print('claim boundary: local active-axis shorthand, not P_t or a chart entry')
fig,ax=plt.subplots()
ax.plot(range(1,len(labels)+1),[d for _,_,d in labels],marker='o')
ax.set_yscale('log'); ax.set_xlabel('prefix length'); ax.set_ylabel('denominator'); ax.set_title('Local active-axis denominator')
plt.show()
```

```python
def P1(x,y): return x*y
def P2(x,y): return 2*x*y
values={'P1(1,1)':P1(1,1),'P2(1,1)':P2(1,1),'same_custody_seed':(0,(1,1),0,0,1,'')}
passed=P1(1,1)!=P2(1,1)
print('PASS' if passed else 'FAIL')
print('exact values:', values)
print('claim boundary: custody-only axioms do not determine a unique pairing seed')
fig,ax=plt.subplots()
ax.bar(['P1(1,1)','P2(1,1)'],[P1(1,1),P2(1,1)])
ax.set_ylabel('pairing value'); ax.set_title('Distinct seeds over identical custody')
plt.show()
```

```python
import math, matplotlib.pyplot as plt
D=12
rows=[(a, D//math.gcd(a,D) if a else 1) for a in range(D)]
single=[a for a,L in rows if L==D]
plt.figure(); plt.bar([a for a,_ in rows],[L for _,L in rows]); plt.xlabel('increment a'); plt.ylabel('cycle length'); plt.title('Translations on Z/12Z'); plt.show()
print('PASS' if single==[1,5,7,11] else 'FAIL', {'single_cycle_count':len(single),'increments':single})
```

```python
claims={'Z12_local':1,'axis_blocks_1_to_2':1,'Z24_local':1,'product_derived':0,'prefix_address_derived':0}
plt.figure(); plt.bar(list(claims),list(claims.values())); plt.xticks(rotation=30,ha='right'); plt.ylim(0,1.2); plt.title('Carrier claim dispositions'); plt.show()
print('PASS' if claims['product_derived']==0 and claims['prefix_address_derived']==0 else 'FAIL', claims)
```

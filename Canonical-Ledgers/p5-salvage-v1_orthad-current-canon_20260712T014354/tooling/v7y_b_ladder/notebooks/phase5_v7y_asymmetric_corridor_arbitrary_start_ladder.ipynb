```python
import math, numpy as np, matplotlib.pyplot as plt
def B(u,v): return v,u+v
def Bk(u,v,k):
    for _ in range(k): u,v=B(u,v)
    return u,v
def fibs(n):
    f=[0,1]
    for _ in range(2,n+3): f.append(f[-1]+f[-2])
    return f
def formula(u,v,k):
    f=fibs(k+2)
    if k==0: return u,v
    return f[k-1]*u+f[k]*v, f[k]*u+f[k+1]*v
starts=[(1,n) for n in range(2,12)] + [(2,13),(3,17),(7,19),(11,4)]
checks=[]
for s in starts:
    for k in range(13): checks.append(Bk(*s,k)==formula(*s,k))
passed=all(checks)
plt.figure(); plt.plot([sum(Bk(*starts[0],k)) for k in range(13)], marker='o'); plt.title('Width growth from arbitrary 1/n start'); plt.xlabel('depth'); plt.ylabel('u+v'); plt.show()
print('PASS' if passed else 'FAIL', {'checks_passed':sum(checks), 'checks_total':len(checks)})
```

```python
def Binv(u,v): return v-u,u
def Binvk(u,v,k):
    for _ in range(k): u,v=Binv(u,v)
    return u,v
recover=[]
for s in starts:
    for k in range(13): recover.append(Binvk(*Bk(*s,k),k)==s)
passed=all(recover)
plt.figure(); plt.bar(range(len(starts)), [sum(Bk(*s,8)) for s in starts]); plt.title('Distinct arbitrary starts at depth 8'); plt.xlabel('start index'); plt.ylabel('u+v'); plt.show()
print('PASS' if passed else 'FAIL', {'inverse_checks_passed':sum(recover), 'checks_total':len(recover)})
```

```python
def wedge(p,q): return p[0]*q[1]-p[1]*q[0]
pairs=[((2,13),(7,19)), ((1,9),(11,4)), ((3,17),(5,22))]
checks=[]; vals=[]
for p,q in pairs:
    base=wedge(p,q)
    for k in range(7):
        obs=wedge(Bk(*p,k), Bk(*q,k)); exp=((-1)**k)*base
        checks.append(obs==exp); vals.append(obs)
passed=all(checks)
plt.figure(); plt.plot(vals, marker='o'); plt.title('Unimodular wedge sign alternation'); plt.xlabel('sample'); plt.ylabel('wedge'); plt.show()
print('PASS' if passed else 'FAIL', {'wedge_checks_passed':sum(checks), 'checks_total':len(checks)})
```

```python
records=[]
for i,s in enumerate(starts):
    for k in range(8):
        u,v=Bk(*s,k); key=(k,(u+v)%12)
        records.append((key,s,(u,v)))
buckets={}
for key,s,t in records: buckets.setdefault(key,[]).append((s,t))
collisions=[(key,b) for key,b in buckets.items() if len({x[0] for x in b})>1]
passed=len(collisions)>0
plt.figure(); plt.bar([0,1], [len(collisions), len(records)]); plt.title('Projection collisions vs total records'); plt.xticks([0,1], ['collisions','records']); plt.ylabel('count'); plt.show()
print('PASS' if passed else 'FAIL', {'projection_collision_witnesses':len(collisions), 'records':len(records)})
```

```python

from fractions import Fraction
from math import gcd, lcm
import matplotlib.pyplot as plt

def ratio(a,b):
    # lens value = (amp, phase mod 4); return b/a
    return (b[0]/a[0], (b[1]-a[1])%4)

def mul(x,y): return (x[0]*y[0], (x[1]+y[1])%4)

def lens(uv, ph): return (Fraction(1,uv), ph%4)
axes=[lens(1,0), lens(6,2), lens(15,1)]
cycle=mul(mul(ratio(axes[0],axes[1]), ratio(axes[1],axes[2])), ratio(axes[2],axes[0]))
passed = cycle == (Fraction(1),0)
plt.figure()
plt.bar(['amp numerator','amp denominator','phase'], [cycle[0].numerator, cycle[0].denominator, cycle[1]])
plt.title('Claim 1: compiler cocycle triangle closes')
plt.show()
print('PASS' if passed else 'FAIL', {'cycle': cycle})
```

```python

def c_pair(Di,Dj,seed):
    L=lcm(Di,Dj); g=gcd(Di,Dj); k=seed%g or 1
    return k*(L//g)%L
Ds=[2,12,30,80]
Cs={(i,j):c_pair(Ds[i],Ds[j],i+2*j+1) for i in range(len(Ds)) for j in range(i+1,len(Ds))}
rep_ok=[]
for (i,j),c in Cs.items():
    L=lcm(Ds[i],Ds[j]); g=gcd(Ds[i],Ds[j])
    rep_ok.append(c%(L//g)==0)
plt.figure()
plt.bar([f'{i}-{j}' for i,j in Cs], list(Cs.values()))
plt.title('Claim 2: T-derived pair coefficients are representative-valid')
plt.show()
print('PASS' if all(rep_ok) else 'FAIL', {'pairs':len(Cs),'valid':sum(rep_ok)})
```

```python

def factor(n):
    d=2; out={}
    while d*d<=n:
        while n%d==0:
            out[d]=out.get(d,0)+1; n//=d
        d += 1 if d==2 else 2
    if n>1: out[n]=out.get(n,0)+1
    return out
primes=sorted({p for d in Ds for p in factor(d)})
counts={p:sum(1 for d in Ds if p in factor(d)) for p in primes}
plt.figure()
plt.bar([str(p) for p in primes], [counts[p] for p in primes])
plt.title('Claim 3: mixed-prime cyclic carrier decomposition is detected')
plt.show()
print('PASS', counts)
```

```python

rank=12
large_D=[2,4,12,30,80,208,12,80,30,4,208,2]
# Nonbruteforce key size proxy: grows by rank and edge summaries, not by product(D).
product=1
for d in large_D: product*=d
key_terms=rank + rank*(rank-1)//2
passed = product > 10**12 and key_terms < 1000
plt.figure()
plt.bar(['log10 group size','key terms'], [len(str(product))-1, key_terms])
plt.title('Claim 4: large-rank path avoids full orbit enumeration')
plt.show()
print('PASS' if passed else 'FAIL', {'rank':rank,'group_size_digits':len(str(product)),'key_terms':key_terms})
```
